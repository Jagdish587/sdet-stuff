```python
# 1. Store the function
res = lambda a, b: a + b

print(res)       # function
print(res(2, 3)) # 5
```

```python
# 2. Immediately execute the lambda
res = (lambda a, b: a + b)(2, 3)

print(res)
```

```python
# 3. You can also use existing variables
a = 2
b = 3

res = (lambda x, y: x + y)(a, b)

print(res) 
```

```python
marks = {
    "John": 85,
    "Alice": 92,
    "Bob": 78,
    "David": 88
}

# Sort by KEY
print(dict(sorted(marks.items(), key=lambda item: item[0])))
# or 
print(sorted(marks))

# Sort by VALUE
print(dict(sorted(marks.items(), key=lambda item: item[1])))

# Descending
print(dict(sorted(marks.items(), key=lambda item: item[1], reverse=True)))
```

```python
# sort based on len , lex order
names = ['Bob', 'John', 'David', 'Alexander']
print(sorted(names, key = lambda name : len(name)))
print(sorted(names, key = lambda name : name))
print(sorted(names, key = lambda name : name, reverse=True))
```

```python
# sort by list indices
mylist = [1, 2, 3,7, 6, 5, 4, 8, 9, 10]
left = 3
right = 6
print("id = ", id(mylist))


mylist[left:right+1] = sorted(mylist[left:right+1])
print(mylist)
print("id = ", id(mylist))
```
